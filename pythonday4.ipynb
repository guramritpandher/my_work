# **RECURSION**

```python
#Technique in programming in which function can call itself to small minr problems:
def recr(x):
  if(x > 0):
    result = x + recr(x - 1)
    print(result)
  else:
    result = 0
  return result

print("Recursion Example Results:")
recr(6)
```

```python
#factorial using python recursion:
def fact(n):
  if n<0:
    print('Sorrryyy')
  elif n==0:
    return 1
  else:
    return n * fact(n-1)

print(fact(5))
```

```python
# Fibonacci Series Using Recursion
def fib(n):
  if n==0:
    return 0
  elif n==1:
    return 1
  else:
    return fib(n-1)+fib(n-2)

print(fib(7))
```

```python
class Atm():

  def __init__(self):
    self.pin=""
    self.balance=0
    self.menu()

  def menu(self):
    user_inpt=input("""
              Hey,Please Proceed according to your need:
              1. Enter 1 to create pin
              2. Enter 2 to deposit
              3. Enter 3 to withdraw
              4. Enter 4 to check balance
              5. Enter 5 to exit
    """)
    if user_inpt=='1':
      self.create_pin()
    elif user_inpt=='2':
      self.deposit()
    elif user_inpt=='3':
      self.withdraw()
    elif user_inpt=='4':
      self.check_balance()
    else:
      print("Bye")

  def create_pin(self):
    self.pin=input("Enter your pin")
    print("Pin Set Successfully")

  def deposit(self):
    dep=input("Enter your pin")
    if dep==self.pin:
      amount=int(input("Enter the amount"))
      self.balance=self.balance+amount
      print("Deposit Successful")
    else:
      print("Invalid Pin")

  def withdraw(self):
    dep=input("Enter your pin")
    if dep==self.pin:
      amount=int(input("Enter the amount"))
      if amount<self.balance:
        self.balance=self.balance-amount
        print("Collect Your Cash")
      else:
        print("Oopss!!Not enough funds")
    else:
      print("Invalid pin")

  def check_balance(self):
    dep=input("Enter Your pin")
    if dep==self.pin:
      print("Your Total Balance: ",self.balance)
    else:
      print("Invalid Pin")
```
