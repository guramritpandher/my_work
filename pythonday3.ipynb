# **DICTIONARY**

```python
# Dictionary: used to store data in key:value pair
#Dictionary constructor: dict()
my_dict={
    'name':'Guramrit',
    'surname':'pandher',
    'age':'23',
    'features':['eyes','lips']
}
print(my_dict)
print(my_dict['name'])
print(len(my_dict))
print(type(my_dict))
```

```python
#ACCESS ITEMS OF DICT:
print(my_dict['age'])

#get() method:
x=my_dict.get("surname")
print(x)

#keys() method: would return list of all the keys in our dict:
x=my_dict.keys()
print(x)

#values() method: would return list of all the values of our dict
x=my_dict.values()
print(x)

#items() method: would return list of all the key values in form of tuples:
x=my_dict.items()
print(x)

if 'name' in my_dict:
  print(f'Yes', {my_dict.get('name')})
```

```python
#Change Dictionary Items:
#to change value for key directly:
my_dict['age']=24
print(my_dict)

#update() method: will change the the mentioned value for specific key
x=my_dict.update({"name":'onal'})
print(my_dict)
```

```python
#Adding Items to dictionary:
#we can add new items using 2 ways:
my_dict['eyes']=2
print(my_dict)

#using update() method:
my_dict.update({'ears':2})
print(my_dict)
```

```python
#Remove Items form dict:
#pop() this method removes the item with specific name:
my_dict.pop('ears')
print(my_dict)

#popitem() this method removes last item of the dict:
my_dict.popitem()
print(my_dict)

#del works to delete specific key, also whole fdictionary
del my_dict['age']
print(my_dict)

del my_dict
print(my_dict)

#clear method helps to empty the dictionary
my_dict.clear()
```

```python
#loop dictionary:
#to get access to keys
dict_1={
    'pencil':['extradark','natraj','doms'],
    'eraser':3,
    'pen':'hauser',
    'pen_color':'black'
}
for i in dict_1:
  print(i)
print('\n')

for i in dict_1.keys():
  print(i)
print('\n')

#to get access to values
for i in dict_1:
  print(dict_1[i])
print('\n')

for i in dict_1.values():
  print(i)
print('\n')

#to get access to items
for i,j in dict_1.items():
  print(i,j)

```

```python
#Copy Dictionary:
# copy() method
new_dict=dict_1.copy()
print(new_dict)

#dict() can be used as well to create a copy\
# new_dict1=dict(dict_1)
# print(new_dict)
```

```python
#Nested DICT:
myfamily = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007
  },
  "child3" : {
    "name" : "Linus",
    "year" : 2011
  }
}
print(myfamily)


#seprate dictionaries and then on more dictionary holding values names of rest of the dictionaries
child1 = {
  "name" : "Emil",
  "year" : 2004
}
child2 = {
  "name" : "Tobias",
  "year" : 2007
}
child3 = {
  "name" : "Linus",
  "year" : 2011
}

myfamily = {
  "child1" : child1,
  "child2" : child2,
  "child3" : child3
}


#acess in nested list:
print(myfamily['child1']['year'])


#loop through nested dict
for i, itm in myfamily.items():
    print(i)

    for j in itm:
        print(j + ':', itm[j])
```

```python
# Write a function that takes two numbers as arguments and returns their sum.
def sum(num1,num2):
  return num1+num2

sum(3,7)
```

```python
# Write a function that takes a string as input and returns the string reversed.
def reverse_string(abc):
  return abc[::-1]

reverse_string('hello')
```

```python
# Write a function that takes a list of numbers and returns the largest number in the list.
def largest_num(xyz):
  a=xyz[0]
  for i in xyz:
    if i>a:
      a=i
  return a

q=largest_num([1,2,3,4,5])
print(q)
```

```python
# Write a function that checks if a given number is even or odd.
def even_odd(num):
  if num%2==0:
    return 'even'
  else:
    return 'odd'

even_odd(3)
```

```python
# Write a function that takes a number as input and returns its factorial.
def fact(num):
  if num<0:
    print('Give number greater than this')
  elif num==0:
    print('Fatorial = 1')
  else:
    factorial = 1
    for i in range(1,num+1):
      factorial=factorial*i
    return factorial

fact(5)
```

```python
#*args:
# function will receive a tuple of arguments
def grade(*clas):
  print("In which class do you study: "+ clas[1])

grade('first','second','third')

# Can also send arguments with the key = value syntax.
def grade(clas1,clas2,clas3):
  print("In which class do u study: "+clas3)

grade(clas1='first',clas2='second',clas3='third')
```

```python
# **Kwargs:
# If the number of keyword arguments is unknown, add a double ** before the parameter name:
def my_function(**kid):
  print("My last name is " + kid["lname"])

my_function(fname = "Onal", lname = "Pandher")
```

```python
#Positional Only argument
def my_function(x, /):
  print(x)

my_function(3)
```

```python
#LAMBDA FUNCTION:
#can take n number of arguments
x = lambda a : a + 10
print(x(5))


x = lambda a, b, c : a + b + c
print(x(1, 2, 3))

def my_fxn(a):
  return lambda b : b * a

inpt = my_fxn(2)

print(inpt(5))
```
